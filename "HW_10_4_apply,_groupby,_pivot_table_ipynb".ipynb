Чоловіки переважають серед незареєстрованих покупців (`Normal`) і мають найнижчий середній чек, оскільки сприймають візит як «швидку покупку» і відмовляються від довгої реєстрації на касі.

**Бізнес-рішення**

* **Instant-реєстрація (без тертя):** QR-код на касі/чеку з можливістю додати картку в Apple/Google Wallet за 15 секунд під час черги.
* **Таргетовані тригери:** Миттєва знижка «тут і зараз» на чоловічі та стабільні категорії (`Electronic accessories`, `Sports and travel`) за умови швидкої реєстрації.
* **Персоналізований Retention:** Push-повідомлення з акціями на товари регулярного вжитку (кава, перекуси, павербанки) для формування звички повертатися.

**Очікуваний ефект**
Переведення чоловіків із категорії `Normal` у `Member` дозволить підтягнути їхній середній чек і частоту покупок до показників жінок-учасниць клубу лояльності.

```python
df['Date'] = pd.to_datetime(df['Date'])


df['Month'] = df['Date'].dt.month_name()        # Назва місяця (January, February...)
df['Month_Num'] = df['Date'].dt.month           # Номер місяця для правильного сортування (1-9)
df['Day_of_Week'] = df['Date'].dt.day_name()    # Назва дня тижня (Monday, Tuesday...)
df['Day_Num'] = df['Date'].dt.dayofweek         # Номер дня тижня (0 = Понеділок, 6 = Неділя)
df['Is_Weekend'] = df['Day_Num'].apply(lambda x: 'Weekend' if x >= 5 else 'Weekday') # Вихідний чи будній
```

```python
weekday_analysis = (
    df.groupby(['Day_Num', 'Day_of_Week', 'Product line'])['Total']
    .agg(Total_Sales='sum', Transaction_Count='count', Avg_Sales='mean')
    .reset_index()
    .sort_values(by=['Day_of_Week', 'Total_Sales'], ascending=[True, False])
    .drop(columns=['Day_Num'])

).round(2)

print("--- Продажі по днях тижня ---")
print(weekday_analysis.to_string(index=False))
```

### Загальний стратегічний висновок

Управління мережею має базуватися на **двох рівнях адаптації**:

1. **Динамічний маркетинг і викладка (Тижневі ритми):**
Маркетингову активність, пропозиції лояльності та торцеві викладки потрібно підлаштовувати під день тижня, закриваючи актуальну потребу споживача саме в день візиту.
2. **Гіперлокалізація та мікроаналіз (Регіональний контекст):**
Універсальний підхід не працює однаково для всіх міст. Необхідно провести **мікроисследования для кожної точки**, щоб адаптувати асортимент під географічні та соціальні особливості:
* **Гірський/туристичний регіон (як Мандалай):** акцент на гірське спорядження, трекінг, прокат інвентарю та захист від сонця.
* **Прибрежний/пляжний регіон:** розширення асортименту пляжних аксесуарів, охолоджених напоїв та швидких перекусів.
* **Адміністративні центри (як Нейпідо):** фокус на готові обіди, ділові аксесуари та корпоративні пропозиції.
* **Мегаполіси (як Янгон):** товари для швидкого облаштування побуту та домашній затишок.



Гнучке поєднання **тижневих поведінкових патернів** із **локальною специфікою регіону** дозволить максимізувати конверсію, підвищити середній чек і забезпечити найвищу віддачу з кожного квадратного метра торговельної площі.
